```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Select a subset of numeric columns relevant to student performance
numeric_cols = ['studytime', 'failures', 'absences', 'G1', 'G2', 'G3']
sub_df = df[numeric_cols]

# 1. Calculate the Covariance Matrix
covariance_matrix = sub_df.cov()
print("=== Covariance Matrix ===")
print(covariance_matrix)
print("\n" + "="*40 + "\n")

# 2. Calculate the Correlation Matrix (Pearson correlation by default)
correlation_matrix = sub_df.corr()
print("=== Correlation Matrix ===")
print(correlation_matrix)

# 3. Visualize the Correlation Matrix using a Seaborn Heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Correlation Matrix of Student Performance Variables')
plt.show()
```
